```python
id = 5
name = "Alice"
age = 30
employer = "TechCorp"
email = "alice@techcorp.com"
salary = 75000.50

print(f"ID: {id}")
print(f"Name: {name}")
print(f"Age: {age}")
print(f"Employer: {employer}")
print(f"Email: {email}")
print(f"Salary: {salary}")

print("Type of ID:", type(id).__name__)
print("Type of Name:", type(name))
print("Type of Age:", type(age))
print("Type of Employer:", type(employer))
print("Type of Email:", type(email))
print("Type of Salary:", type(salary))
```
